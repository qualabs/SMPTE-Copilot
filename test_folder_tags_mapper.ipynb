# Test FolderTagsMapper

This notebook tests the `FolderTagsMapper` class to verify it correctly maps file paths to access tags based on their parent folder.


```python
import sys
import importlib.util
from pathlib import Path

# Import directly from file to avoid importing src/__init__.py 
# which has dependencies we don't need for testing
spec = importlib.util.spec_from_file_location(
    "folder_tags_mapper",
    Path.cwd() / "src" / "config" / "folder_tags_mapper.py"
)
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)
FolderTagsMapper = module.FolderTagsMapper

print("✓ FolderTagsMapper imported successfully!")
```

## Test 1: Create mapper with mapping file


```python
# Create mapper instance
mapper = FolderTagsMapper(
    mapping_file=Path("folder_tags_mapping.json"),
    default_tags=["Public"]
)

print(f"Mapper created with default tags: {mapper.default_tags}")
print(f"Mapping file: {mapper.mapping_file}")
```

## Test 2: Test files in mapped folders


```python
# Test file in Finance folder (should return ["Finance", "Public"])
file = Path("data/Finance/25_10_14__R1_TMulagada_Man_Final.pdf")
tags = mapper.get_tags_for_file(file)
print(f"\nFile: {file}")
print(f"Tags: {tags}")
```

## Test 3: Test file in unmapped folder (should use default tags)


```python
# Test file in unmapped folder (should return default_tags: ["Public"])
other_file = Path("data/h264_small.pdf")
tags = mapper.get_tags_for_file(other_file)
print(f"\nFile: {other_file}")
print(f"Tags: {tags}")
```
